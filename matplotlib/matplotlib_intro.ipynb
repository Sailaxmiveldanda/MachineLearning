## Anatomy of Matplotlib

```python
x = [1,2,3,4]
y = [11,22,33,44]
fig,axes = plt.subplots(figsize = (10,10)) #width, height
axes.plot(x,y)
axes.set(title = "title", xlabel = "x", ylabel = "y")
fig.savefig("sample_plot.png")
```

## Types of Matplotlib
1. Line Plot - ax.plot() - ax.linspace()
2. Scatter - ax.scatter()
3. bar plot - ax.bar()
4. histogram - ax.hist()

```python
import numpy as np
```

```python
x = np.linspace(0,10,100)
x[:5]
```

```python
fig,ax = plt.subplots()
ax.plot(x, x ** 2)
```

```python
fig,ax = plt.subplots()
ax.scatter(x, np.exp(x))
```

```python
fig,ax = plt.subplots()
ax.scatter(x, np.sin(x))
```

```python
fig,ax = plt.subplots()
ax.bar(x, np.std(x))
```

```python
data = {"Almond" : 10,
       "Cashew" : 15,
       "Peanut" : 8}
fig,ax = plt.subplots()
ax.bar(data.keys(), data.values())
ax.set(title = "Store", ylabel = "Price")
```

```python
fig,ax = plt.subplots()
ax.barh(data.keys(), data.values())
# ax.set(title = "Store", ylabel = "Price")
```

```python
v = np.random.rand(1000)
fig,ax = plt.subplots()
ax.hist(v);
```

```python
p = np.random.random(1000)
fig,ax = plt.subplots()
ax.hist(p);
```

```python
fig, ((ax1,ax2), (ax3,ax4)) = plt.subplots(nrows= 2, ncols=2, figsize=(5,7))
ax1.plot(x, x/2)
ax1.set_title("Line Plot")
ax2.scatter(np.random.random(10), np.random.random(10))
ax3.bar(data.keys(), data.values())
ax4.hist(np.random.random(100))
ax2.set_title("Scatter Plot")
ax3.set_title("Bar Plot")
ax4.set_title("Hist Plot")
```

```python
fig, ax = plt.subplots(nrows= 2, ncols=2, figsize=(5,7))
ax[0][0].plot(x, x/2);
ax[0][0].set_title("Line Plot")
ax[0][1].scatter(np.random.random(10), np.random.random(10));
ax[1][0].bar(data.keys(), data.values());
ax[1][1].hist(np.random.random(100));
ax[0][1].set_title("Scatter Plot")
ax[1][0].set_title("Bar Plot")
ax[1][1].set_title("Hist Plot")
```
