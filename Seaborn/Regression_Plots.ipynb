```python
import seaborn as sns
```

```python
# lmplot - display linear models, split the plots based on features
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
%matplotlib inline
```

```python
sns.Implot(x = 'total_bill', y = 'tip', data = tips)
```

```python
import matplotlib.pyplot as plt
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips)
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, hue = 'sex')
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, hue = 'sex', palette = 'coolwarm')
```

## Working with Markers
markers - control of shape of points 
scatter_kws = scatter plot apperance (Customize)

```python
sns.lmplot(x = 'total_bill', y = 'tip', hue = 'sex', data = tips, palette = 'coolwarm',
           markers = ['o','v'], scatter_kws={'s' : 100})
```

## Using a Grid

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, col = 'sex')
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, row = 'sex')
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, row = 'sex', col = 'time')
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, hue = 'sex', col = 'day', palette = 'coolwarm')
```

## Aspect and height

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, col = 'day', hue = 'sex', palette = 'coolwarm',
          aspect=0.6, height= 8)
```
