## Categorical Plots
- factorplot - relationships across categories
- boxplot - distribution(spread, median,shape)
- violinplot - distribution(spread, median,shape)
- stripplot - individual data points
- swarmplot - individual data points
- barplot - aggregated values
- countplot - aggregated values

```python
import seaborn as sns
```

```python
%matplotlib inline
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

# barplot - allow you to aggregate categorical data based off some function

```python
sns.barplot(x = 'sex', y = 'total_bill', data = tips)
```

```python
import numpy as np
```

```python
#estimator - convert vector to scalar
sns.barplot(x = 'sex', y = 'total_bill', data = tips, estimator = np.std)
```

## countplot - same as barplot except the estimator is explicitly counting the no. of occurances

```python
sns.countplot(x = 'sex', data = tips)
```

```python
sns.countplot(x = 'smoker', data = tips)
```

```python
## box plot 
```

```python
tips.head()
```

```python
sns.boxplot(x = 'day', y = 'total_bill', data = tips, palette = 'rainbow')
```

```python
sns.boxplot(x = 'smoker', y = 'total_bill', data = tips, palette = 'rainbow')
```

```python
sns.boxplot(data=tips,palette = 'rainbow', orient = 'v')
```

```python
sns.boxplot(x = 'day', y = 'total_bill', data = tips, palette = 'coolwarm', hue = 'smoker')
```

```python
## voilin plot
```

```python
sns.violinplot(x = 'day', y = 'total_bill', data=tips, palette = 'rainbow')
```

```python
sns.violinplot(x='day',y='total_bill',data=tips,hue='smoker',palette = 'coolwarm')
```

```python
sns.violinplot(x='day',y='total_bill',data=tips,hue='smoker',palette = 'Set1', split = True)
```

```python
#stripplot
```

```python
sns.stripplot(x='day',y='total_bill',data=tips)
```

```python
sns.stripplot(x='day',y='total_bill',data=tips, jitter = True)
```

```python
sns.stripplot(x='day',y='total_bill',data=tips, hue = 'sex', jitter = True, palette = 'Set1')
```

```python
## swarmplot
```

```python
sns.swarmplot(x='day',y='total_bill', data = tips)
```

```python
sns.swarmplot(x='day',y='total_bill', data = tips,palette= 'Set1', hue = 'sex')
```

## Combine 

```python
sns.violinplot(x='tip',y='day',data = tips, palette = 'rainbow')
sns.swarmplot(x='tip',y='day',data = tips,color = 'black', size = 3)
```

```python
sns.violinplot(x="tip", y="day", data=tips,palette='coolwarm')
sns.swarmplot(x="tip", y="day", data=tips,color='black',size=3)
```
