## Legends Labels Titles

```python
#title - set_title()
# label - x_label(), y_label()
# label - label = label_text, and thenuse legend without arguments
```

```python
fig = plt.figure()
axes = fig.add_axes([0.1,0.1,0.8,0.8])
axes.plot(x, x ** 4, label = "x ** 4")
axes.plot(y, y ** 2, label = "y ** 4")
axes.legend()
```

```python
fig = plt.figure()
axes = fig.add_axes([0.1,0.1,0.8,0.8])
axes.plot(x, x ** 4, label = "x ** 4")
axes.plot(y, y ** 2, label = "y ** 4")
axes.legend(loc=0)
```

```python
fig = plt.figure()
axes = fig.add_axes([0.1,0.1,0.8,0.8])
axes.plot(x, x ** 4, label = "x ** 4")
axes.plot(y, y ** 2, label = "y ** 4")
axes.legend(loc=10)
```

```python
fig = plt.figure()
axes = fig.add_axes([0.1,0.1,0.8,0.8])
axes.plot(x, x ** 4, label = "x ** 4")
axes.plot(y, y ** 2, label = "y ** 4")
axes.legend(loc=(0.1,0.1))
```
