## Types of Matplotlib
1. Line Plot - ax.plot() - ax.linspace()
2. Scatter - ax.scatter()
3. bar plot - ax.bar()
4. histogram - ax.hist()

```python
over_50 = heart_disease[heart_disease["age"] > 50]
over_50
```

```python
over_50.plot(kind = "scatter", 
             x = 'age', y = 'chol', c = 'target', figsize=(10,6)) 
```

```python
fig,ax = plt.subplots(figsize = (10,6))
over_50.plot(kind = 'scatter',
             x = 'age', y = 'chol', c = 'target', ax = ax)
ax.set_xlim([45,100])
```

```python
fig, ax = plt.subplots(figsize = (10,6))
scatter = ax.scatter(x = over_50["age"], y = over_50["chol"], 
                     c = over_50["target"])
ax.set(title = "Heart Disease and Cholestrol Levels", 
       xlabel = "Age", ylabel="Chol")
ax.legend(*scatter.legend_elements(), title = "Target")
```

```python
fig, ax = plt.subplots(figsize = (10,6))
scatter = ax.scatter(x = over_50["age"], y = over_50["chol"], 
                     c = over_50["target"])
ax.set(title = "Heart Disease and Cholestrol Levels", 
       xlabel = "Age", ylabel="Chol")
ax.legend(*scatter.legend_elements(), title = "Target")
ax.axhline(over_50["chol"].mean(), ls = "--")
```

```python
over_50
```

```python
fig, (ax0,ax1) = plt.subplots(nrows=2, ncols=1, figsize=(10,8), sharex= True)
sactter = ax0.scatter(over_50["age"], over_50["chol"], c = over_50["target"])
ax0.set(title = "Heart Disease and Cholestrol Levels", xlabel = "x", ylabel = "Cholestrol")
ax0.legend(*scatter.legend_elements(), title = "Target")
ax0.axhline(y = over_50["chol"].mean(), ls = "--", color = 'g', label = "Average")
```

```python
fig, (ax0,ax1) = plt.subplots(nrows=2, ncols=1, figsize=(10,8), sharex= True)
sactter = ax0.scatter(over_50["age"], over_50["chol"], c = over_50["target"])
ax0.set(title = "Heart Disease and Cholestrol Levels", xlabel = "Age", ylabel = "Cholestrol")
ax0.legend(*scatter.legend_elements(), title = "Target")
ax0.axhline(y = over_50["chol"].mean(), ls = "--", color = 'g', label = "Average")

sactter = ax1.scatter(over_50["age"], over_50["thalach"], c = over_50["target"])
ax1.set(title = "Heart Disease and Max Heart Rate Levels", xlabel = "Age", ylabel = "Max Heart Rate")
ax1.legend(*scatter.legend_elements(), title = "Target")
ax1.axhline(y = over_50["thalach"].mean(), ls = "--", color = 'r', label = "Average")

fig.suptitle("Heart Disease Analysis", fontsize = 15, fontweight = 'bold')
```

```python
fig,ax = plt.subplots(figsize = (10,6))
scatter = ax.scatter(x = over_50["age"], y = over_50["chol"], c = over_50["target"], cmap = "viridis")

ax.set(title="Heart Disease and Cholesterol Levels",
       xlabel="Age",
       ylabel="Cholesterol");
ax.axhline(y = over_50["chol"].mean(), c= 'b', ls = "--", label =  "Average")
ax.legend(*scatter.legend_elements(), title = "Target")
```

```python
fig,ax = plt.subplots(figsize = (10,6))
scatter = ax.scatter(x = over_50["age"], y = over_50["chol"], c = over_50["target"], cmap = "winter")

ax.set(title="Heart Disease and Cholesterol Levels",
       xlabel="Age",
       ylabel="Cholesterol");
ax.axhline(y = over_50["chol"].mean(), c= 'b', ls = "--", label =  "Average")
ax.legend(*scatter.legend_elements(), title = "Target")
```

```python
fig, (ax0, ax1) = plt.subplots(nrows=2, 
                               ncols=1, 
                               sharex=True, 
                               figsize=(10, 7))
scatter = ax0.scatter(over_50["age"], 
                      over_50["chol"], 
                      c=over_50["target"],
                      cmap="winter")
ax0.set(title="Heart Disease and Cholesterol Levels",
        ylabel="Cholesterol")
ax0.axhline(y=over_50["chol"].mean(), 
            color="r", 
            linestyle="--", 
            label="Average");
ax0.legend(*scatter.legend_elements(), title="Target")
scatter = ax1.scatter(over_50["age"], 
                      over_50["thalach"], 
                      c=over_50["target"],
                      cmap="winter")
ax1.set(title="Heart Disease and Max Heart Rate Levels",
        xlabel="Age",
        ylabel="Max Heart Rate")
ax1.axhline(y=over_50["thalach"].mean(), 
            color="r", 
            linestyle="--", 
            label="Average");
ax1.legend(*scatter.legend_elements(), 
           title="Target")
fig.suptitle("Heart Disease Analysis", 
             fontsize=16, 
             fontweight="bold");
```

```python
fig, (ax0, ax1) = plt.subplots(nrows=2, 
                               ncols=1, 
                               sharex=True, 
                               figsize=(10, 7))
scatter = ax0.scatter(over_50["age"], 
                      over_50["chol"], 
                      c=over_50["target"],
                      cmap="winter")
ax0.set(title="Heart Disease and Cholesterol Levels",
        ylabel="Cholesterol", xlim = [50,80])
ax0.axhline(y=over_50["chol"].mean(), 
            color="r", 
            linestyle="--", 
            label="Average");
ax0.legend(*scatter.legend_elements(), title="Target")
scatter = ax1.scatter(over_50["age"], 
                      over_50["thalach"], 
                      c=over_50["target"],
                      cmap="winter")
ax1.set(title="Heart Disease and Max Heart Rate Levels",
        xlabel="Age",
        ylabel="Max Heart Rate", ylim = [60,220])
ax1.axhline(y=over_50["thalach"].mean(), 
            color="r", 
            linestyle="--", 
            label="Average");
ax1.legend(*scatter.legend_elements(), 
           title="Target")
fig.suptitle("Heart Disease Analysis", 
             fontsize=16, 
             fontweight="bold");
```
